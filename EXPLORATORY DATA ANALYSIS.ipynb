```python

# Step 10.2: IQR Outlier Detection

# 1. Check the Amount column
print("Amount data type:", df["Amount"].dtype)
print("Missing Amount values:", df["Amount"].isna().sum())
print("Total rows:", len(df))

# 2. Calculate quartiles
Q1 = df["Amount"].quantile(0.25)
Q3 = df["Amount"].quantile(0.75)

# 3. Calculate IQR
IQR = Q3 - Q1

# 4. Calculate outlier boundaries
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# 5. Identify potential outliers
outlier_mask = (
    (df["Amount"] < lower_bound) |
    (df["Amount"] > upper_bound)
)

outliers = df.loc[outlier_mask]

# 6. Calculate count and percentage
outlier_count = len(outliers)
total_rows = len(df)

outlier_percentage = (
    outlier_count / total_rows * 100
    if total_rows > 0 else 0
)

# 7. Display results
print("\nQ1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)
print("Potential outliers:", outlier_count)
print(f"Outlier percentage: {outlier_percentage:.2f}%")

# 8. View the first 10 flagged transactions
print(outliers[["Time", "Amount", "Class"]].head(10))
```
